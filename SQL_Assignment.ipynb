## Q NO 2 : Explain the purpose of constraints and how they help maintain data integrity in a database. Provide examples of common types of constraints.


## ANS:
###Purpose of Constraints in a Database :

Constraints in a database are rules or conditions applied to the data in tables to ensure its accuracy, consistency, and reliability, collectively referred to as **data integrity**. They enforce specific requirements on how data can be inserted, updated, or deleted, preventing invalid or inconsistent data from being stored. Constraints help maintain data integrity by:

**1. Ensuring Data Accuracy**: Constraints prevent incorrect or invalid data entries, such as ensuring a numeric field only contains numbers.

 **2. Maintaining Consistency**: They ensure relationships between tables remain valid, such as ensuring a foreign key references an existing record.

 **3. Preventing Data Corruption**: Constraints protect the database from operations that could violate its logical structure or rules.

 **4. Enforcing Business Rules**: Constraints can enforce specific business logic, like ensuring an employee's hire date is not in the future.

**5. Supporting Data Reliability**: By enforcing rules, constraints ensure the database remains a trustworthy source for applications and users.

### How Constraints Help Maintain Data Integrity:

Data integrity refers to the accuracy, completeness, and consistency of data. Constraints contribute to different types of data integrity:

**Entity Integrity:** Ensures each row in a table is unique and identifiable (e.g., no duplicate rows).

**Referential Integrity:** Ensures relationships between tables are valid (e.g., foreign keys reference existing primary keys).

**Domain Integrity:** Ensures data values in a column meet specific criteria (e.g., valid data types or ranges).

**User-Defined Integrity:** Enforces business-specific rules (e.g., a column must contain a specific format).

By enforcing these rules, constraints prevent issues like orphaned records, invalid data entries, or violations of business logic, ensuring the database remains reliable for querying and reporting.

### Common Types of Constraints with Examples:

Below are the most common types of constraints used in relational databases, along with examples:

**1. Primary Key Constraint:**

Purpose: Ensures each row in a table is uniquely identifiable and prevents duplicate or null values in the primary key column(s).

Example: In a `Students` table, a `student_id` column is set as the primary key to uniquely identify each student.

E.g:  

    CREATE TABLE Students (
          student_id INT PRIMARY KEY,
          name VARCHAR(50),
          email VARCHAR(100)
    );

This ensures `student_id` is unique and not null for every record.



 **2. Foreign Key Constraint:**

Purpose: Ensures that a value in a column (or set of columns) matches a value in a referenced table’s primary key or unique key, maintaining referential integrity.

Example: In an `Enrollments` table, a `student_id` column references the `student_id` in the `Students` table.

E.g:

      CREATE TABLE Enrollments (
          enrollment_id INT PRIMARY KEY,
          student_id INT,
          course_id INT,
          FOREIGN KEY (student_id) REFERENCES Students(student_id)
      );

 This ensures `student_id` in `Enrollments` must exist in the `Students` table, preventing orphaned records.

**3. Unique Constraint:**

Purpose: Ensures all values in a column or set of columns are unique across rows, but allows null values (unlike a primary key).

Example: In a `Users` table, the `email` column must be unique to prevent duplicate email addresses.

E.g:

    CREATE TABLE Users (
          user_id INT PRIMARY KEY,
          email VARCHAR(100) UNIQUE,
          username VARCHAR(50)
    );

This ensures no two users can have the same email address, but `email` can be null if allowed.

**4. Not Null Constraint:**

Purpose: Ensures a column cannot contain null values, enforcing that every row must have a valid value for that column.

Example: In an `Employees` table, the `name` column must always have a value.

E.g:

      CREATE TABLE Employees (
          employee_id INT PRIMARY KEY,
          name VARCHAR(50) NOT NULL,
          hire_date DATE
      );

This ensures every employee record must include a name.

**5. Check Constraint:**

Purpose: Ensures values in a column meet a specific condition or range, enforcing domain integrity.

Example: In a `Products` table, the `price` column must be greater than zero.

E.g:

      CREATE TABLE Products (
          product_id INT PRIMARY KEY,
          name VARCHAR(100),
          price DECIMAL(10,2) CHECK (price > 0)
      );

 This ensures no product can have a negative or zero price.

**6. Default Constraint:**

Purpose: Specifies a default value for a column if no value is provided during an insert operation.

Example: In an `Orders` table, the `order_date` column defaults to the current date if not specified.
E.g:

CREATE TABLE Orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    order_date DATE DEFAULT current_timestamp
);

This ensures that if no `order_date` is provided, the current time is automatically inserted.














## Q NO 3: Why would you apply the NOT NULL constraint to a column? Can a primary key contain NULL values? Justify your answer.


##ANS:  
###Why apply the NOT NULL constraint?

The NOT NULL constraint ensures that a column must always have a value.


It prevents storing NULL (unknown or missing values) in that column.


This is useful when a field is mandatory for every record.


✅ Example:
    
    emp_name VARCHAR(100) NOT NULL

This makes sure every employee record has a name.

###Can a Primary Key contain NULL values?

No. A Primary Key cannot contain NULL values.

Justification:

**1.Uniquely identifies each row**


- A primary key must uniquely identify a record in a table.


- If it allowed NULL, then a row with NULL wouldn’t have a unique identifier.


**2.Primary Key = UNIQUE + NOT NULL**


- In SQL, defining a column as PRIMARY KEY automatically applies both UNIQUE and NOT NULL constraints.


**3.Relational Integrity**


- Other tables often use FOREIGN KEYS to reference the primary key.


If the key were NULL, referential integrity would break (you can’t reference “nothing”).


✅ Example:

    CREATE TABLE employees (
        emp_id INT PRIMARY KEY,
        emp_name VARCHAR(100) NOT NULL
    );

Here, `emp_id` cannot be NULL and must be unique.


## Q NO 4 :  Explain the steps and SQL commands used to add or remove constraints on an existing table. Provide an example for both adding and removing a constraint.

##ANS :
### Managing Constraints on an Existing Table

Constraints can be added or removed on an already existing table using the ALTER TABLE command.


###✅ 1. Adding a Constraint

Syntax:

    ALTER TABLE table_name
    ADD CONSTRAINT constraint_name constraint_type (column_name);

**Example – Add a UNIQUE constraint**

Suppose we have a table employees without a constraint on the email column:

    CREATE TABLE employees (
        emp_id INT PRIMARY KEY,
        emp_name VARCHAR(100) NOT NULL,
        email VARCHAR(255),
        age INT
    );

**Add a UNIQUE constraint on email:**

    ALTER TABLE employees
    ADD CONSTRAINT unique_email UNIQUE (email);

**Add a CHECK constraint to ensure age ≥ 18:**

    ALTER TABLE employees
    ADD CONSTRAINT check_age CHECK (age >= 18);


### ✅ 2. Removing a Constraint

Syntax:
    
    ALTER TABLE table_name
    DROP CONSTRAINT constraint_name;

- In MySQL, the keyword is `DROP INDEX` for UNIQUE constraints, and `DROP FOREIGN KEY` for foreign keys. For SQL Server/Oracle/PostgreSQL, `DROP CONSTRAINT` works.

**Example – Remove a Constraint**

**Drop the UNIQUE constraint from email:**

    ALTER TABLE employees
    DROP INDEX unique_email;   -- MySQL

**Drop the CHECK constraint on age:**

    ALTER TABLE employees
    DROP CONSTRAINT check_age;   -- SQL Server / PostgreSQL / Oracle



## Q NO 5 :  Explain the consequences of attempting to insert, update, or delete data in a way that violates constraints. Provide an example of an error message that might occur when violating a constraint.

## ANS :

###Consequences of Violating Constraints

### 1. NOT NULL Constraint
- Rule: Column cannot be left empty.


- Violation: Inserting or updating a row without a value.


- Consequence: Error → prevents incomplete/invalid records.


Example:


    INSERT INTO employees (emp_id, emp_name, email, age)
    VALUES (1, NULL, 'abc@mail.com', 25);

❌ **Error (MySQL):**

- ERROR 1048 (23000): Column `emp_name` cannot be null



###2. PRIMARY KEY Constraint

- Rule: Must be unique and NOT NULL.


- Violation: Inserting a duplicate or NULL key.


- Consequence: Prevents duplicate or unidentified rows.


Example:


    INSERT INTO employees (emp_id, emp_name, email, age)
    VALUES (1, 'Alice', 'alice@mail.com', 30);

    INSERT INTO employees (emp_id, emp_name, email, age)
    VALUES (1, 'Bob', 'bob@mail.com', 28);  -- duplicate ID

❌ **Error:**

- ERROR 1062 (23000): Duplicate entry '1' for key 'PRIMARY'


###3. UNIQUE Constraint

- Rule: No duplicate values in that column.


- Violation: Inserting or updating with a duplicate.


- Consequence: Prevents duplicate data.


Example:


    INSERT INTO employees (emp_id, emp_name, email, age)
    VALUES (2, 'Charlie', 'alice@mail.com', 29); -- duplicate email

❌ **Error:**

- ERROR 1062 (23000): Duplicate entry 'alice@mail.com' for key 'unique_email'


###4. CHECK Constraint

- Rule: Value must meet a condition.


- Violation: Insert/update outside the allowed range.


- Consequence: Prevents invalid data values.


Example:


    INSERT INTO employees (emp_id, emp_name, email, age)
    VALUES (3, 'David', 'david@mail.com', 15); -- too young

❌ **Error:**

- ERROR 3819 (HY000): Check constraint 'check_age' is violated.


###5. FOREIGN KEY Constraint

- Rule: Value must exist in the parent table.


- Violation: Insert orphan record, or delete a parent row in use.


- Consequence: Protects referential integrity.


Example:


    INSERT INTO orders (order_id, emp_id)
    VALUES (101, 99);  -- emp_id 99 does not exist in employees

❌ **Error:**

- ERROR 1452 (23000): Cannot add or update a child row:
a foreign key constraint fails (`orders`, CONSTRAINT `fk_emp` FOREIGN KEY (`emp_id`) REFERENCES `employees` (`emp_id`))

### Differences Between Primary Keys and Foreign Keys

While both primary keys (PKs) and foreign keys (FKs) are crucial for database structure and integrity, they serve distinct purposes and have key differences:

1. **Purpose**:
   - **PK**: Uniquely identifies each record in its own table, ensuring no duplicates and serving as the main identifier for rows (e.g., `actor_id` in `actor` uniquely identifies an actor).
   - **FK**: Establishes a link between two tables by referencing a PK in another table, enforcing relationships and data consistency across tables (e.g., `customer_id` in `payment` links to `customer(customer_id)` to associate payments with customers).

2. **Uniqueness and NULL Values**:
   - **PK**: Must be unique and cannot contain NULL values. A table can have only one PK (which may be composite, like in `film_actor`).
   - **FK**: Does not need to be unique (multiple rows can reference the same PK value) and can contain NULL values unless a `NOT NULL` constraint is applied. A table can have multiple FKs.

3. **Indexing**:
   - **PK**: Automatically creates a unique index for fast lookups and is often used as the clustered index in storage.
   - **FK**: Typically indexed for performance but not necessarily unique. The index helps enforce referential integrity checks during inserts/updates.

4. **Referential Role**:
   - **PK**: Is the "target" of references; other tables point to it.
   - **FK**: Is the "pointer"; it references a PK or unique key in another table. Actions like `ON DELETE` or `ON UPDATE` (e.g., CASCADE, RESTRICT) define how changes in the referenced table affect the FK table.

5. **Enforcement and Integrity**:
   - **PK**: Ensures entity integrity (each row is distinct).
   - **FK**: Ensures referential integrity (no "orphan" records; e.g., you can't add a rental for a non-existent customer). Violations trigger errors.

6. **Examples in Context**:
   - In this database, PKs like `film_id` in `film` ensure each movie is unique, while FKs like `film_id` in `inventory` link inventory items to specific films, preventing invalid references.

### Step 1: Identify a Table in the Sakila Database that Violates 3NF
The Sakila database is generally well-normalized, but the **address** table violates 3NF due to a transitive dependency involving the `district` column.

**address Table Schema**:

- Primary Key: `address_id`
- Purpose: Stores address information for customers, staff, and stores, with `city_id` linking to the `city` table.

The table is in 1NF (atomic values, unique rows) and 2NF (single-column PK, no partial dependencies). However, it violates 3NF due to transitive dependencies.

### Step 2: Describe the Transitive Dependencies Present
The primary key is `address_id` (single column), so all non-key attributes should depend directly on `address_id`. However, the following transitive dependency exists:

- `address_id` → `city_id` → `district`
  - `city_id` is a non-key attribute (it's a foreign key, but not part of the PK).
  - `district` (e.g., state, province, or region) is determined by `city_id`, not directly by `address_id`.
  - In the Sakila data, each `city_id` corresponds to a specific city in a specific district. For example:
    - All addresses with the same `city_id` (e.g., city_id = 300, city = "London") have the same `district` (e.g., "Ontario" or whatever the data shows; in Sakila, districts are like provinces/states).
    - This means `district` depends on `city_id` (since a city is located in one district), creating a transitive dependency.
  - This leads to redundancy: If multiple addresses share the same `city_id`, the `district` value is repeated for each address, rather than stored once in the `city` table.
  - Potential anomalies:
    - **Update Anomaly**: If a city's district changes (e.g., due to boundary changes), you must update `district` in every address row for that `city_id`, risking inconsistency.
    - **Insertion Anomaly**: You can't add a new city without an address, but the district is tied to addresses.
    - **Deletion Anomaly**: Deleting all addresses for a city could lose the district information for that city.

Other potential transitive dependencies in `address` could involve `postal_code` or `phone`, but they are more address-specific and less clearly transitive. The `district` → `city_id` is the clearest violation based on logical dependencies in the data.

(Note: To verify this in the actual data, you can run a query like:
```sql
SELECT city_id, district, COUNT(*) AS address_count
FROM address
GROUP BY city_id, district
HAVING COUNT(DISTINCT district) > 1;
```
If this returns no rows, it confirms that each `city_id` has only one `district`, proving the transitive dependency. In standard Sakila data, this is the case.)

### Step 3: Outline the Steps to Normalize the Table to 3NF
To achieve 3NF, eliminate the transitive dependency by moving the dependent attribute (`district`) to the table where it belongs (the `city` table, since `city_id` determines `district`). This decomposes the table while preserving data and dependencies.

#### Normalization Steps:
1. **Identify the Transitive Dependency and New Table Structure**:
   - Move `district` to the `city` table, as `city_id` (PK of `city`) determines `district`.
   - The `city` table already exists in Sakila: `city_id` (PK), `city` (VARCHAR(50)), `country_id` (FK), `last_update` (TIMESTAMP).
   - Add `district` to `city` to resolve the dependency.

     ```sql
     ALTER TABLE city
     ADD COLUMN district VARCHAR(20) NOT NULL DEFAULT '';
     ```

       ```sql
       UPDATE city c
       SET c.district = (
           SELECT district
           FROM address a
           WHERE a.city_id = c.city_id
           LIMIT 1  -- Assuming consistency
       );
       ```
     - If there are inconsistencies (rare in Sakila), resolve them by choosing the most common `district` or manually correcting.

   - **Remove `district` from the address table**:
     ```sql
     ALTER TABLE address
     DROP COLUMN district;
     ```
     - This eliminates the transitive dependency, as `district` is now accessed via `city_id` → `city.district`.

3. **Verify Data Integrity and Relationships**:
   - Ensure foreign keys and indexes remain intact.
   - Update any views, stored procedures, or queries that referenced `address.district` to join with `city` instead (e.g., `SELECT a.*, c.district FROM address a JOIN city c ON a.city_id = c.city_id`).
   - Test for lossless decomposition: The original data can be reconstructed by joining `address` and `city` on `city_id`.

4. **Resulting Normalized Tables**:
   - **address table** (now in 3NF):
     | address_id | address | address2 | city_id | postal_code | phone | location | last_update |
     |------------|---------|----------|---------|-------------|-------|----------|-------------|
     - All non-key attributes now depend directly on `address_id` (no transitive dependencies).
   - **city table** (updated, still in 3NF):

## a. Take a specific table in Sakila and guide through the process of normalizing it from the initial unnormalized form up to at least 2NF.


## ANS:

### Step 1: Hypothesize an Unnormalized Customer Table
In its standard form, the Sakila `customer` table is already normalized:

  - `store_city` and `store_manager` depend on `store_id`, not directly on `customer_id`, suggesting a partial or transitive dependency.
- **Repeating Data**: `store_city` and `store_manager` are repeated for customers from the same store, causing redundancy.

### Step 2: Normalize to First Normal Form (1NF)
**1NF Requirements**:
- All attributes must contain atomic (indivisible) values.
- No repeating groups or arrays in a single column.
- Each row must be unique (enforced by a primary key).

**Violations in the Unnormalized Table**:
- `phone_numbers` contains multiple values (e.g., "123-456-7890,987-654-3210"), violating atomicity.
- `customer_name` combines first and last names, which can be split into separate atomic attributes.
- The table has a primary key (`customer_id`), so row uniqueness is satisfied.

**Normalization Steps for 1NF**:
1. **Split Non-Atomic Columns**:
   - Split `customer_name` into `first_name` and `last_name` to make them atomic.
   - Move `phone_numbers` to a separate table to handle multiple phone numbers as individual rows.
2. **Create a New Table for Phone Numbers**:
   - Create a `customer_phone` table to store each phone number in a separate row, linked to `customer_id`.
3. **Retain Other Attributes**: For now, keep `store_id`, `store_city`, `store_manager`, etc., as we’ll address dependencies in 2NF.

**Resulting Tables**:
- **customer table** (temporary, still needs 2NF):
  | customer_id | first_name | last_name | store_id | store_city | store_manager | email                        | address_id | active | create_date         |

  | customer_id | phone_number     |

**SQL to Create 1NF Tables**:
```sql
-- Create the modified customer table (still needs 2NF)
CREATE TABLE customer (
    customer_id SMALLINT UNSIGNED NOT NULL,
    first_name VARCHAR(45) NOT NULL,
    last_name VARCHAR(45) NOT NULL,
    store_id SMALLINT UNSIGNED NOT NULL,
    store_city VARCHAR(50) NOT NULL,
    store_manager VARCHAR(50) NOT NULL,
    email VARCHAR(50) DEFAULT NULL,
    address_id SMALLINT UNSIGNED NOT NULL,
    active TINYINT(1) NOT NULL DEFAULT 1,
    create_date DATETIME NOT NULL,
    last_update TIMESTAMP DEFAULT CURRENT_TIMESTAMP ON UPDATE CURRENT_TIMESTAMP,
    PRIMARY KEY (customer_id),
    FOREIGN KEY (store_id) REFERENCES store(store_id),
    FOREIGN KEY (address_id) REFERENCES address(address_id)
);

-- Create the customer_phone table
CREATE TABLE customer_phone (
    customer_id SMALLINT UNSIGNED NOT NULL,
    phone_number VARCHAR(20) NOT NULL,
    PRIMARY KEY (customer_id, phone_number),
    FOREIGN KEY (customer_id) REFERENCES customer(customer_id)
);
```

### Step 3: Normalize to Second Normal Form (2NF)
**2NF Requirements**:
- The table must be in 1NF.
- All non-key attributes must be fully functionally dependent on the entire primary key (no partial dependencies).

**Analysis of the Customer Table**:
- **Primary Key**: `customer_id` (single column).
- **Non-Key Attributes**: `first_name`, `last_name`, `store_id`, `store_city`, `store_manager`, `email`, `address_id`, `active`, `create_date`, `last_update`.
- Since the primary key is a single column (`customer_id`), there can be no partial dependencies (partial dependencies occur only with composite keys). Thus, a table with a single-column primary key is automatically in 2NF if it is in 1NF.
- However, the `customer` table has a **transitive dependency** (relevant for 3NF, but we’ll address it here to show the process):
  - `store_id` → `store_city`, `store_manager` (non-key attributes `store_city` and `store_manager` depend on `store_id`, not directly on `customer_id`).
  - This is a 3NF violation, but for completeness, we’ll move these attributes to achieve a fully normalized structure.

To ensure 2NF (and prepare for 3NF), we’ll address the logical structure by moving store-related attributes to the `store` table (which already exists in Sakila) and keeping only `store_id` in `customer`.

**Normalization Steps for 2NF (and Beyond)**:
1. **Identify Partial/Transitive Dependencies**:
   - `store_city` and `store_manager` depend on `store_id`, not directly on `customer_id`. These should reside in the `store` table, as `store_id` determines them.

2. **Move Store-Related Attributes**:
   - The Sakila database already has a `store` table: `store (store_id, manager_staff_id, address_id, last_update)`.
   - In a real-world scenario, `store_city` and `store_manager` would belong in the `store` table (or related tables like `address` and `staff`). For simplicity, assume:
     - `store_city` is derived from `store.address_id` → `address.city_id` → `city.city`.
     - `store_manager` is derived from `store.manager_staff_id` → `staff.first_name`, `staff.last_name`.
   - Remove `store_city` and `store_manager` from the `customer` table, as they are redundant and can be retrieved via joins with `store`, `staff`, and `address`.

3. **Update the Customer Table**:
   - The `customer` table should only include `store_id` as a foreign key to reference the `store` table.
   - Resulting `customer` table schema:
     | customer_id | first_name | last_name | store_id | email                        | address_id | active | create_date         | last_update         |
     |-------------|------------|-----------|----------|------------------------------|------------|--------|---------------------|---------------------|
     | 1           | MARY       | SMITH     | 1        | MARY.SMITH@sakilacustomer.org | 5          | 1      | 2006-02-14 22:04:36 | 2006-02-15 04:57:20 |
     | 2           | PATRICIA   | JOHNSON   | 2        | PATRICIA.JOHNSON@sakilacustomer.org | 6    | 1      | 2006-02-14 22:04:36 | 2006-02-15 04:57:20 |
     | 3           | LINDA      | WILLIAMS  | 1        | LINDA.WILLIAMS@sakilacustomer.org | 7      | 1      | 2006-02-14 22:04:36 | 2006-02-15 04:57:20 |

4. **Update the Store Table (if needed)**:
   - The Sakila `store` table already exists, but for completeness, ensure it can support queries for `store_city` and `store_manager` via joins:
     - `store_city` is in `city` (via `store.address_id` → `address.city_id` → `city.city`).
     - `store_manager` is in `staff` (via `store.manager_staff_id` → `staff.first_name`, `staff.last_name`).
   - No changes are needed to `store`, as it already supports these relationships.

**SQL to Modify the Customer Table**:
```sql
-- Create the normalized customer table
CREATE TABLE customer (
    customer_id SMALLINT UNSIGNED NOT NULL,
    first_name VARCHAR(45) NOT NULL,
    last_name VARCHAR(45) NOT NULL,
    store_id SMALLINT UNSIGNED NOT NULL,
    email VARCHAR(50) DEFAULT NULL,
    address_id SMALLINT UNSIGNED NOT NULL,
    active TINYINT(1) NOT NULL DEFAULT 1,
    create_date DATETIME NOT NULL,
    last_update TIMESTAMP DEFAULT CURRENT_TIMESTAMP ON UPDATE CURRENT_TIMESTAMP,
    PRIMARY KEY (customer_id),
    FOREIGN KEY (store_id) REFERENCES store(store_id),
    FOREIGN KEY (address_id) REFERENCES address(address_id)
);

-- The customer_phone table (from 1NF) remains unchanged
```

## a. Write a query using a CTE to retrieve the distinct list of actor names and the number of films they have acted in from the actor and film_actor tables.


## ANS:

```sql
WITH ActorFilmCounts AS (
    SELECT
        a.actor_id,
        a.first_name,
        a.last_name,
        COUNT(fa.film_id) AS film_count
    FROM actor a
    JOIN film_actor fa ON a.actor_id = fa.actor_id
    GROUP BY a.actor_id, a.first_name, a.last_name
)
SELECT
    first_name,
    last_name,
    film_count
FROM ActorFilmCounts
ORDER BY last_name, first_name;
```

## a. Write a query using a CTE to find the total revenue generated by each customer (sum of payments) from the customer and payment tables.

## ANS:

```sql
WITH CustomerRevenue AS (
    SELECT
        c.customer_id,
        c.first_name,
        c.last_name,
        SUM(p.amount) AS total_revenue
    FROM customer c
    JOIN payment p ON c.customer_id = p.customer_id
    GROUP BY c.customer_id, c.first_name, c.last_name
)
SELECT
    first_name,
    last_name,
    total_revenue
FROM CustomerRevenue
ORDER BY total_revenue DESC, last_name, first_name;
```

## a. Utilize a CTE with a window function to rank films based on their rental duration from the film table.


## ANS:

```sql
WITH FilmRentalDuration AS (
    SELECT
        f.film_id,
        f.title,
        AVG(DATEDIFF(r.return_date, r.rental_date)) AS avg_rental_duration
    FROM film f
    LEFT JOIN inventory i ON f.film_id = i.film_id
    LEFT JOIN rental r ON i.inventory_id = r.inventory_id
    WHERE r.return_date IS NOT NULL
    GROUP BY f.film_id, f.title
)
SELECT
    title,
    avg_rental_duration,
    RANK() OVER (ORDER BY avg_rental_duration DESC) AS duration_rank
FROM FilmRentalDuration
ORDER BY duration_rank, title;
```

## a. Create a CTE to list customers who have made more than two rentals, and then join this CTE with the customer table to retrieve additional customer details.


## ANS:


```sql
WITH FrequentRenters AS (
    SELECT
        c.customer_id,
        COUNT(r.rental_id) AS rental_count
    FROM customer c
    JOIN rental r ON c.customer_id = r.customer_id
    GROUP BY c.customer_id
    HAVING COUNT(r.rental_id) > 2
)
SELECT
    c.customer_id,
    c.first_name,
    c.last_name,
    c.email,
    c.active,
    fr.rental_count
FROM FrequentRenters fr
JOIN customer c ON fr.customer_id = c.customer_id
ORDER BY fr.rental_count DESC, c.last_name, c.first_name;
```

## a. Write a query using a CTE to find the total number of rentals made each month, considering the rental_date from the rental table.





## ANS:

```sql
WITH MonthlyRentals AS (
    SELECT
        YEAR(rental_date) AS year,
        MONTH(rental_date) AS month,
        COUNT(rental_id) AS rental_count
    FROM rental
    GROUP BY YEAR(rental_date), MONTH(rental_date)
)
SELECT
    year,
    month,
    rental_count
FROM MonthlyRentals
ORDER BY year, month;
```

## a. Create a CTE to generate a report showing pairs of actors who have appeared in the same film together, using the film_actor table.


## ANS:


```sql
WITH ActorPairs AS (
    SELECT
        fa1.actor_id AS actor_id_1,
        fa2.actor_id AS actor_id_2,
        fa1.film_id
    FROM film_actor fa1
    JOIN film_actor fa2 ON fa1.film_id = fa2.film_id AND fa1.actor_id < fa2.actor_id
)
SELECT
    a1.first_name AS actor1_first_name,
    a1.last_name AS actor1_last_name,
    a2.first_name AS actor2_first_name,
    a2.last_name AS actor2_last_name,
    f.title AS film_title
FROM ActorPairs ap
JOIN actor a1 ON ap.actor_id_1 = a1.actor_id
JOIN actor a2 ON ap.actor_id_2 = a2.actor_id
JOIN film f ON ap.film_id = f.film_id
ORDER BY f.title, a1.last_name, a1.first_name, a2.last_name, a2.first_name;
```

## a. Implement a recursive CTE to find all employees in the staff table who report to a specific manager, considering the reports_to column.


## ANS:

```sql
WITH RECURSIVE StaffHierarchy AS (
    -- Anchor member: direct reports of the manager
    SELECT
        s.staff_id,
        s.first_name,
        s.last_name,
        s.reports_to
    FROM staff s
    WHERE s.reports_to = 1   -- starting manager

    UNION ALL

    -- Recursive member: employees who report to staff in previous level
    SELECT
        s.staff_id,
        s.first_name,
        s.last_name,
        s.reports_to
    FROM staff s
    INNER JOIN StaffHierarchy sh
        ON s.reports_to = sh.staff_id
)
SELECT *
FROM StaffHierarchy;
```
